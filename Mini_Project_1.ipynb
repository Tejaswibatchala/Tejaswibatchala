Task-2

```python
import pandas as pd
import numpy as np
```

```python
#Remove duplicate rows
df = df.drop_duplicates()
```

```python
#Strip extra spaces from the fspecified columns:
column_to_strip=["student_name","topic","tool_used"]
for col in column_to_strip:
  if col in df.columns:
    df[col]=df[col].astype(str).str.strip()

```

```python
#Convert the following columns to title case:
column_to_title_case=["attendance","lab_completed","api_used"]
for col in column_to_title_case:
  if col in df.columns:
    df[col]=df[col].astype(str).str.title()

```

```python
#Replace invalid assignment_score values
df.loc[
    (df["assignment_score"]<0)|(df["assignment_score"]>100),
    "assignment_score"
]=np.nan
```

```python
#Replace invalid quiz_score values
df.loc[
    (df['quiz_score']<0)|(df['quiz_score']>100),
    "quiz_score"
]=np.nan
```

```python
#Replace invalid study_hours values:
df.loc[
    (df["study_hours"]<0),
    "study_hours"
]=np.nan
```

```python
#Replace invalid feedback_rating values:
df.loc[
    (df["feedback_rating"]<1)|(df["feedback_rating"]>5),
    "feedback_rating"
]=np.nan
```

```python
#Fill missing values
df["assignment_score"].fillna(df["assignment_score"].mean(),inplace=True)
df["quiz_score"].fillna(df["quiz_score"].mean(),inplace=True)
df["study_hours"].fillna(df["study_hours"].median(),inplace=True)
df["feedback_rating"].fillna(df["feedback_rating"].median(),inplace=True)

```

```python
print("shape of the cleaned DataFram:")
print(df.shape)
```

```python
#Print the missing value count after cleaning:
print("Print the missing value count after cleaning:")
print(df.isnull().sum())
```

```python
#Save the cleaned dataset:
df.to_csv("cleaned_ai_learning_lab.csv",index=False)
print("/ncleaned dataset as'cleaned_ai_learning_lab.csv'")
```

```python
import pandas as pd
import numpy as np
df=pd.read_csv("cleaned_ai_learning_lab.csv")
#Create a new column called total_score
df["total_score"]=df["assignment_score"]+df["quiz_score"]
#Create a  performance_level column
def get_performance_level(score):
  if score>=160:
    return "Excellent"
  elif score>=120:
    return "Good"
  elif score>=80:
    return "Average"
  else:
    return "Needs support"
df["performance_level"]=df["total_score"].apply(get_performance_level)
#Print the average assignment score
print("average assignment score:")
print(df["assignment_score"].mean())
#Print the average quiz score.
print("average quiz score:")
print(df["quiz_score"].mean())
#Print the average study hours
print("average study hours:")
print(df["study_hours"].mean())
#Print the number of students in each performance_level
print("/nstudents in each performance_level:")
print(df["performance_level"].value_counts())
#Print the average score topic-wise using groupby()
print("/n total score topic-wise:")
print(df.groupby("topic")["total_score"].mean())
#Print the average score batch-wise using groupby()
print("/n total score batch-wise:")
print(df.groupby("batch")["total_score"].mean())
#Print students who are marked as Needs Support
print("\nStudents Marked as Needs Support:")
needs_support = df[df["performance_level"] == "Needs Support"]

if "student_name" in df.columns:
    print(needs_support[["student_name", "total_score", "performance_level"]])
else:
    print(needs_support)

```
